```python
# ======================================================================
# CREDIT RISK MODEL FOR PRIVATE DEBT / DIRECT LENDING
# FINAL ONE-CELL GOOGLE COLAB PROJECT
# ======================================================================
#
# PROJECT PURPOSE
# ----------------
# This project demonstrates an end-to-end credit underwriting workflow:
#
# Historical corporate financial data
#            ↓
# Financial-ratio engineering
#            ↓
# Probability-of-default model
#            ↓
# Public-company borrower scoring
#            ↓
# Hypothetical direct-lending underwriting
#            ↓
# Pricing / facility sizing / covenants
#            ↓
# Stress testing
#            ↓
# Final credit memo
#
# IMPORTANT:
# The historical bankruptcy training population and current SEC
# public-company population are different populations and time periods.
# Therefore, the resulting PD is a MODEL-IMPLIED RESEARCH ESTIMATE,
# not a production-calibrated U.S. private-credit PD.
#
# THIS VERSION ALSO:
# - Does NOT install or upgrade NumPy/Pandas
# - Does NOT use SciPy
# - Does NOT use scikit-learn
# - Does NOT use fit_transform()
# - Handles SEC gzip/compression safely
# - Uses multiple SEC XBRL fallback tags
# - Validates SEC data before scoring
# - Uses default_flag instead of SQL-reserved "default"
# - Uses only three meaningful visualizations
# - Creates a professional Excel credit memo
# - Creates CSV outputs
# - Creates a SQLite database
# - Creates a saved model artifact
# - Produces a beginning company overview
# - Produces a final closing credit analysis
#
# ======================================================================


# ======================================================================
# 1. IMPORTS
# ======================================================================

import os
import io
import json
import csv
import time
import gzip
import zipfile
import sqlite3
import urllib.request
import urllib.error
import importlib.util
from pathlib import Path

# ----------------------------------------------------------------------
# Only install genuinely missing lightweight packages.
# Never upgrade Colab's existing NumPy/Pandas installation.
# ----------------------------------------------------------------------

required_packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "openpyxl": "openpyxl",
}

missing_packages = []

for module_name, package_name in required_packages.items():

    if importlib.util.find_spec(module_name) is None:

        missing_packages.append(
            package_name
        )

if missing_packages:

    import subprocess

    subprocess.check_call(
        [
            "pip",
            "install",
            "--quiet",
            *missing_packages,
        ]
    )

import numpy as np
import pandas as pd

from openpyxl import Workbook
from openpyxl.styles import (
    Font,
    PatternFill,
    Alignment,
)
from openpyxl.utils import get_column_letter

import matplotlib.pyplot as plt


# ======================================================================
# 2. GLOBAL CONFIGURATION
# ======================================================================

SEED = 42

np.random.seed(
    SEED
)

rng = np.random.default_rng(
    SEED
)

PROJECT_ROOT = (
    Path.cwd()
    /
    "credit-risk-direct-lending"
)

RAW_DIR = (
    PROJECT_ROOT
    /
    "data"
    /
    "raw"
)

OUTPUT_DIR = (
    PROJECT_ROOT
    /
    "outputs"
)

CHART_DIR = (
    OUTPUT_DIR
    /
    "charts"
)

MODEL_DIR = (
    PROJECT_ROOT
    /
    "models"
)

SQL_DIR = (
    PROJECT_ROOT
    /
    "sql"
)

for directory in [
    RAW_DIR,
    OUTPUT_DIR,
    CHART_DIR,
    MODEL_DIR,
    SQL_DIR,
]:

    directory.mkdir(
        parents=True,
        exist_ok=True,
    )


# ======================================================================
# 3. PROJECT INTRODUCTION
# ======================================================================

PROJECT_TITLE = (
    "Credit Risk Model for Private Debt / Direct Lending"
)

PROJECT_DESCRIPTION = """
This project builds a complete credit-risk and direct-lending
underwriting framework.

The system first learns historical relationships between corporate
financial ratios and bankruptcy outcomes. It then applies the resulting
model to a selected public-company borrower using SEC financial data.

The final underwriting layer converts the model-implied probability of
default into a hypothetical direct-lending analysis covering:

• probability of default
• illustrative spread and coupon
• maximum facility size
• leverage covenant
• interest coverage covenant
• expected loss
• risk-adjusted yield
• EBITDA and interest-rate stress testing

The purpose is to demonstrate the workflow used in quantitative credit
research and private-credit underwriting.

IMPORTANT LIMITATION:
The historical bankruptcy dataset and the current SEC public-company
population are not identical populations. Consequently, the model-
implied PD should be treated as a research estimate and not as a
production-calibrated private-credit PD.
"""

# ----------------------------------------------------------------------
# Companies
# ----------------------------------------------------------------------

COMPANY_UNIVERSE = {

    "MSFT": {
        "name":
            "Microsoft Corporation",

        "sector":
            "Technology",

        "description":
            (
                "Global technology company with businesses spanning "
                "software, cloud computing, enterprise technology "
                "and digital services."
            ),
    },

    "AAPL": {
        "name":
            "Apple Inc.",

        "sector":
            "Technology",

        "description":
            (
                "Technology company that develops consumer devices, "
                "software and digital services."
            ),
    },

    "AMZN": {
        "name":
            "Amazon.com, Inc.",

        "sector":
            "Consumer / Technology",

        "description":
            (
                "Global company operating across e-commerce, cloud "
                "computing, AI, digital services and logistics."
            ),
    },

    "WMT": {
        "name":
            "Walmart Inc.",

        "sector":
            "Retail",

        "description":
            (
                "Global omnichannel retailer operating stores, clubs "
                "and e-commerce channels."
            ),
    },
}

PRIMARY_BORROWER = "MSFT"


# ======================================================================
# 4. INTRODUCTORY OUTPUT
# ======================================================================

print("\n" + "=" * 90)
print(
    PROJECT_TITLE.upper()
)
print("=" * 90)

print(
    PROJECT_DESCRIPTION.strip()
)

print(
    "\n"
    + "-" * 90
)

print(
    "SAMPLE COMPANIES USED IN THE ANALYSIS"
)

print(
    "-" * 90
)

company_overview_df = pd.DataFrame(
    [
        {
            "Ticker":
                ticker,

            "Company":
                company["name"],

            "Sector":
                company["sector"],

            "What the Company Does":
                company["description"],
        }

        for ticker, company
        in COMPANY_UNIVERSE.items()
    ]
)

display(
    company_overview_df
)

print(
    "\nPrimary borrower used for the detailed hypothetical "
    "direct-lending case:"
)

print(
    f"{COMPANY_UNIVERSE[PRIMARY_BORROWER]['name']} "
    f"({PRIMARY_BORROWER})"
)

print(
    COMPANY_UNIVERSE[
        PRIMARY_BORROWER
    ][
        "description"
    ]
)


# ======================================================================
# 5. DATA SOURCES
# ======================================================================

UCI_URL = (
    "https://archive.ics.uci.edu/static/public/365/"
    "polish%2Bcompanies%2Bbankruptcy%2Bdata.zip"
)

SEC_TICKERS_URL = (
    "https://www.sec.gov/files/company_tickers.json"
)

SEC_USER_AGENT = os.getenv(
    "SEC_USER_AGENT",
    "CreditRiskDirectLending/1.0 analyst@example.com",
)


# ======================================================================
# 6. SELECTED MODEL FEATURES
# ======================================================================

MODEL_FEATURES = [

    "Attr1",
    "Attr2",
    "Attr3",
    "Attr4",
    "Attr6",
    "Attr7",
    "Attr8",
    "Attr9",
    "Attr10",
    "Attr19",
    "Attr23",
    "Attr27",
    "Attr28",
    "Attr29",
    "Attr30",
    "Attr38",
    "Attr42",
    "Attr50",
    "Attr51",
    "Attr56",
    "Attr59",
]

FEATURE_LABELS = {

    "Attr1":
        "Net Profit / Total Assets",

    "Attr2":
        "Total Liabilities / Total Assets",

    "Attr3":
        "Working Capital / Total Assets",

    "Attr4":
        "Current Assets / Short-Term Liabilities",

    "Attr6":
        "Retained Earnings / Total Assets",

    "Attr7":
        "EBIT / Total Assets",

    "Attr8":
        "Equity / Total Liabilities",

    "Attr9":
        "Sales / Total Assets",

    "Attr10":
        "Equity / Total Assets",

    "Attr19":
        "Gross Profit / Sales",

    "Attr23":
        "Net Profit / Sales",

    "Attr27":
        "Operating Profit / Interest Expense",

    "Attr28":
        "Working Capital / Fixed Assets",

    "Attr29":
        "Log Total Assets",

    "Attr30":
        "(Liabilities - Cash) / Sales",

    "Attr38":
        "Constant Capital / Total Assets",

    "Attr42":
        "Operating Profit / Sales",

    "Attr50":
        "Current Assets / Total Liabilities",

    "Attr51":
        "Current Liabilities / Total Assets",

    "Attr56":
        "Gross Margin",

    "Attr59":
        "Long-Term Liabilities / Equity",
}


# ======================================================================
# 7. GENERAL HELPERS
# ======================================================================

def safe_float(
    value,
    default=np.nan,
):

    try:

        number = float(
            value
        )

        if np.isfinite(
            number
        ):

            return number

        return default

    except Exception:

        return default


def safe_divide(
    numerator,
    denominator,
):

    numerator = safe_float(
        numerator
    )

    denominator = safe_float(
        denominator
    )

    if not np.isfinite(
        numerator
    ):

        return np.nan

    if not np.isfinite(
        denominator
    ):

        return np.nan

    if abs(
        denominator
    ) < 1e-12:

        return np.nan

    return (
        numerator
        /
        denominator
    )


def format_pct(
    value,
):

    value = safe_float(
        value
    )

    if not np.isfinite(
        value
    ):

        return "N/A"

    return (
        f"{value:.2%}"
    )


def format_multiple(
    value,
):

    value = safe_float(
        value
    )

    if not np.isfinite(
        value
    ):

        return "N/A"

    return (
        f"{value:.2f}x"
    )


def format_money(
    value,
):

    value = safe_float(
        value
    )

    if not np.isfinite(
        value
    ):

        return "N/A"

    return (
        f"${value:,.1f}mm"
    )


def download_url(
    url,
    user_agent=None,
    timeout=120,
):

    headers = {}

    if user_agent:

        headers[
            "User-Agent"
        ] = user_agent

    request = (
        urllib.request.Request(
            url,
            headers=headers,
        )
    )

    with urllib.request.urlopen(
        request,
        timeout=timeout,
    ) as response:

        return response.read()


# ======================================================================
# 8. DOWNLOAD UCI DATA
# ======================================================================

print("\n" + "=" * 90)
print(
    "STEP 1 — HISTORICAL BANKRUPTCY DATA"
)
print("=" * 90)

uci_zip_path = (
    RAW_DIR
    /
    "polish_companies_bankruptcy.zip"
)

try:

    if not uci_zip_path.exists():

        print(
            "Downloading UCI bankruptcy dataset..."
        )

        uci_content = download_url(
            UCI_URL,
            timeout=180,
        )

        uci_zip_path.write_bytes(
            uci_content
        )

        print(
            "Download completed."
        )

    else:

        print(
            "Existing UCI dataset found."
        )

except Exception as error:

    raise RuntimeError(
        "Unable to download the UCI dataset.\n"
        f"Error: {error}"
    )


# ======================================================================
# 9. ARFF PARSER
# ======================================================================

def parse_arff(
    raw_bytes,
):

    text = raw_bytes.decode(
        "utf-8",
        errors="replace",
    )

    attributes = []
    rows = []

    reading_data = False

    for raw_line in (
        text.splitlines()
    ):

        line = (
            raw_line.strip()
        )

        if not line:
            continue

        if line.startswith("%"):
            continue

        lower_line = (
            line.lower()
        )

        if lower_line.startswith(
            "@attribute"
        ):

            remainder = (
                line[
                    len(
                        "@attribute"
                    ):
                ]
                .strip()
            )

            if remainder.startswith(
                "'"
            ):

                ending = (
                    remainder.find(
                        "'",
                        1,
                    )
                )

                if ending > 0:

                    name = (
                        remainder[
                            1:
                            ending
                        ]
                    )

                else:

                    name = (
                        remainder.split()[0]
                    )

            elif remainder.startswith(
                '"'
            ):

                ending = (
                    remainder.find(
                        '"',
                        1,
                    )
                )

                if ending > 0:

                    name = (
                        remainder[
                            1:
                            ending
                        ]
                    )

                else:

                    name = (
                        remainder.split()[0]
                    )

            else:

                name = (
                    remainder.split()[0]
                )

            attributes.append(
                name
            )

        elif lower_line.startswith(
            "@data"
        ):

            reading_data = True

        elif reading_data:

            try:

                parsed = next(
                    csv.reader(
                        [line],
                        skipinitialspace=True,
                    )
                )

                if (
                    len(parsed)
                    ==
                    len(attributes)
                ):

                    rows.append(
                        parsed
                    )

            except Exception:

                continue

    if not attributes:

        raise ValueError(
            "ARFF parser found no attributes."
        )

    if not rows:

        raise ValueError(
            "ARFF parser found no observations."
        )

    dataframe = pd.DataFrame(
        rows,
        columns=attributes,
    )

    for column in dataframe.columns:

        dataframe[
            column
        ] = (
            dataframe[
                column
            ]
            .astype(str)
            .str.strip()
            .str.strip("'")
            .str.strip('"')
        )

        dataframe[
            column
        ] = (
            dataframe[
                column
            ].replace(
                {
                    "?": np.nan,
                    "": np.nan,
                    "NA": np.nan,
                    "N/A": np.nan,
                    "nan": np.nan,
                }
            )
        )

    return dataframe


# ======================================================================
# 10. LOAD THREE-YEAR BANKRUPTCY DATA
# ======================================================================

with zipfile.ZipFile(
    uci_zip_path,
    "r",
) as archive:

    archive_names = (
        archive.namelist()
    )

    arff_candidates = [
        name
        for name
        in archive_names
        if name.lower().endswith(
            "3year.arff"
        )
    ]

    if not arff_candidates:

        arff_candidates = [
            name
            for name
            in archive_names
            if (
                "3year"
                in name.lower()
            )
            and
            name.lower().endswith(
                ".arff"
            )
        ]

    if not arff_candidates:

        raise FileNotFoundError(
            "The UCI 3-year ARFF file was not found."
        )

    arff_file = (
        arff_candidates[0]
    )

    arff_bytes = archive.read(
        arff_file
    )

historical_df = parse_arff(
    arff_bytes
)

target_candidates = [
    column
    for column
    in historical_df.columns
    if str(column).lower()
    in {
        "class",
        "target",
        "bankrupt",
        "bankruptcy",
    }
]

if target_candidates:

    target_column = (
        target_candidates[0]
    )

else:

    target_column = (
        historical_df.columns[-1]
    )

historical_df = (
    historical_df
    .rename(
        columns={
            target_column:
                "default_flag"
        }
    )
)

for feature in MODEL_FEATURES:

    historical_df[
        feature
    ] = pd.to_numeric(
        historical_df[
            feature
        ],
        errors="coerce",
    )

historical_df[
    "default_flag"
] = (
    historical_df[
        "default_flag"
    ]
    .astype(str)
    .str.strip()
    .str.strip("'")
    .str.strip('"')
)

historical_df[
    "default_flag"
] = pd.to_numeric(
    historical_df[
        "default_flag"
    ],
    errors="coerce",
)

historical_df = (
    historical_df
    .dropna(
        subset=[
            "default_flag"
        ]
    )
    .copy()
)

historical_df[
    "default_flag"
] = (
    historical_df[
        "default_flag"
    ]
    .astype(int)
)

missing_features = [
    feature
    for feature in MODEL_FEATURES
    if feature
    not in historical_df.columns
]

if missing_features:

    raise ValueError(
        "Missing model features: "
        +
        str(
            missing_features
        )
    )

print(
    f"Historical observations: "
    f"{len(historical_df):,}"
)

print(
    f"Historical default rate: "
    f"{format_pct(historical_df['default_flag'].mean())}"
)


# ======================================================================
# 11. DATA CLEANING
# ======================================================================

print("\n" + "=" * 90)
print(
    "STEP 2 — DATA CLEANING & FEATURE PREPARATION"
)
print("=" * 90)

X_all = (
    historical_df[
        MODEL_FEATURES
    ]
    .copy()
)

X_all = X_all.replace(
    [
        np.inf,
        -np.inf,
    ],
    np.nan,
)

y_all = (
    historical_df[
        "default_flag"
    ]
    .astype(int)
    .to_numpy()
)

print(
    "Cleaning completed."
)


# ======================================================================
# 12. STRATIFIED SPLIT
# ======================================================================

def stratified_split(
    target,
    train_fraction=0.70,
    validation_fraction=0.15,
    seed=42,
):

    target = np.asarray(
        target
    )

    generator = (
        np.random.default_rng(
            seed
        )
    )

    train_indices = []
    validation_indices = []
    test_indices = []

    for class_value in (
        np.unique(target)
    ):

        indices = np.where(
            target
            ==
            class_value
        )[0]

        generator.shuffle(
            indices
        )

        count = len(
            indices
        )

        train_count = int(
            count
            *
            train_fraction
        )

        validation_count = int(
            count
            *
            validation_fraction
        )

        train_indices.extend(
            indices[
                :train_count
            ]
        )

        validation_indices.extend(
            indices[
                train_count:
                train_count
                +
                validation_count
            ]
        )

        test_indices.extend(
            indices[
                train_count
                +
                validation_count:
            ]
        )

    generator.shuffle(
        train_indices
    )

    generator.shuffle(
        validation_indices
    )

    generator.shuffle(
        test_indices
    )

    return (
        np.asarray(
            train_indices,
            dtype=int,
        ),

        np.asarray(
            validation_indices,
            dtype=int,
        ),

        np.asarray(
            test_indices,
            dtype=int,
        ),
    )


train_idx, validation_idx, test_idx = (
    stratified_split(
        y_all,
        seed=SEED,
    )
)

X_train_raw = (
    X_all.iloc[
        train_idx
    ].copy()
)

X_validation_raw = (
    X_all.iloc[
        validation_idx
    ].copy()
)

X_test_raw = (
    X_all.iloc[
        test_idx
    ].copy()
)

y_train = (
    y_all[
        train_idx
    ]
)

y_validation = (
    y_all[
        validation_idx
    ]
)

y_test = (
    y_all[
        test_idx
    ]
)

print(
    f"Training observations:   {len(train_idx):,}"
)

print(
    f"Validation observations: {len(validation_idx):,}"
)

print(
    f"Test observations:       {len(test_idx):,}"
)


# ======================================================================
# 13. ROBUST PREPROCESSING
# ======================================================================

def fit_preprocessor(
    dataframe,
):

    df = dataframe.copy()

    medians = (
        df
        .median()
        .replace(
            [
                np.inf,
                -np.inf,
            ],
            np.nan,
        )
        .fillna(0.0)
    )

    df = (
        df
        .fillna(
            medians
        )
    )

    lower = (
        df
        .quantile(
            0.01
        )
    )

    upper = (
        df
        .quantile(
            0.99
        )
    )

    df = df.clip(
        lower=lower,
        upper=upper,
        axis=1,
    )

    scales = (
        df.quantile(
            0.75
        )
        -
        df.quantile(
            0.25
        )
    )

    scales = (
        scales
        .replace(
            0,
            1.0,
        )
        .replace(
            [
                np.inf,
                -np.inf,
            ],
            np.nan,
        )
        .fillna(1.0)
    )

    return {

        "medians":
            medians,

        "lower":
            lower,

        "upper":
            upper,

        "scales":
            scales,
    }


def transform_preprocessor(
    dataframe,
    parameters,
):

    df = dataframe.copy()

    medians = (
        parameters[
            "medians"
        ]
    )

    lower = (
        parameters[
            "lower"
        ]
    )

    upper = (
        parameters[
            "upper"
        ]
    )

    scales = (
        parameters[
            "scales"
        ]
    )

    df = (
        df
        .replace(
            [
                np.inf,
                -np.inf,
            ],
            np.nan,
        )
        .fillna(
            medians
        )
    )

    df = df.clip(
        lower=lower,
        upper=upper,
        axis=1,
    )

    df = (
        df
        -
        medians
    )

    df = (
        df
        /
        scales
    )

    df = (
        df
        .replace(
            [
                np.inf,
                -np.inf,
            ],
            np.nan,
        )
        .fillna(0.0)
    )

    return (
        df
        .astype(float)
        .to_numpy()
    )


preprocessor = (
    fit_preprocessor(
        X_train_raw
    )
)

X_train = (
    transform_preprocessor(
        X_train_raw,
        preprocessor,
    )
)

X_validation = (
    transform_preprocessor(
        X_validation_raw,
        preprocessor,
    )
)

X_test = (
    transform_preprocessor(
        X_test_raw,
        preprocessor,
    )
)


# ======================================================================
# 14. LOGISTIC REGRESSION
# ======================================================================

class LogisticRegressionNumpy:

    def __init__(
        self,
        learning_rate=0.025,
        iterations=2500,
        l2=0.10,
    ):

        self.learning_rate = (
            learning_rate
        )

        self.iterations = (
            iterations
        )

        self.l2 = l2

        self.weights = None
        self.loss_history = []

    @staticmethod
    def sigmoid(
        values,
    ):

        values = np.clip(
            values,
            -50,
            50,
        )

        return (
            1
            /
            (
                1
                +
                np.exp(
                    -values
                )
            )
        )

    def fit(
        self,
        X,
        y,
    ):

        X = np.asarray(
            X,
            dtype=float,
        )

        y = np.asarray(
            y,
            dtype=float,
        )

        X_design = np.column_stack(
            [
                np.ones(
                    len(X)
                ),
                X,
            ]
        )

        positive_count = max(
            1,
            int(
                np.sum(
                    y == 1
                )
            ),
        )

        negative_count = max(
            1,
            int(
                np.sum(
                    y == 0
                )
            ),
        )

        positive_weight = (
            len(y)
            /
            (
                2
                *
                positive_count
            )
        )

        negative_weight = (
            len(y)
            /
            (
                2
                *
                negative_count
            )
        )

        sample_weights = np.where(
            y == 1,
            positive_weight,
            negative_weight,
        )

        self.weights = np.zeros(
            X_design.shape[1]
        )

        self.loss_history = []

        previous_loss = np.inf

        for _ in range(
            self.iterations
        ):

            scores = (
                X_design
                @
                self.weights
            )

            probabilities = (
                self.sigmoid(
                    scores
                )
            )

            probabilities = np.clip(
                probabilities,
                1e-12,
                1 - 1e-12,
            )

            cross_entropy = (
                -np.average(
                    (
                        y
                        *
                        np.log(
                            probabilities
                        )
                        +
                        (
                            1
                            -
                            y
                        )
                        *
                        np.log(
                            1
                            -
                            probabilities
                        )
                    ),
                    weights=sample_weights,
                )
            )

            regularization = (
                self.l2
                *
                np.sum(
                    self.weights[1:] ** 2
                )
                /
                len(y)
            )

            loss = (
                cross_entropy
                +
                regularization
            )

            self.loss_history.append(
                loss
            )

            error = (
                probabilities
                -
                y
            )

            weighted_error = (
                error
                *
                sample_weights
            )

            gradient = (
                X_design.T
                @
                weighted_error
                /
                np.sum(
                    sample_weights
                )
            )

            gradient[1:] += (
                2
                *
                self.l2
                *
                self.weights[1:]
                /
                len(y)
            )

            self.weights -= (
                self.learning_rate
                *
                gradient
            )

            if (
                abs(
                    previous_loss
                    -
                    loss
                )
                <
                1e-7
            ):

                break

            previous_loss = loss

        return self

    def predict_proba(
        self,
        X,
    ):

        if self.weights is None:

            raise RuntimeError(
                "Logistic model has not been fitted."
            )

        X = np.asarray(
            X,
            dtype=float,
        )

        X_design = np.column_stack(
            [
                np.ones(
                    len(X)
                ),
                X,
            ]
        )

        probability = (
            self.sigmoid(
                X_design
                @
                self.weights
            )
        )

        return np.column_stack(
            [
                1 - probability,
                probability,
            ]
        )


# ======================================================================
# 15. GRADIENT BOOSTING
# ======================================================================

class GradientBoostingStumps:

    def __init__(
        self,
        n_estimators=60,
        learning_rate=0.05,
        max_thresholds=6,
        min_leaf=30,
    ):

        self.n_estimators = (
            n_estimators
        )

        self.learning_rate = (
            learning_rate
        )

        self.max_thresholds = (
            max_thresholds
        )

        self.min_leaf = (
            min_leaf
        )

        self.base_score = None
        self.stumps = []

    @staticmethod
    def sigmoid(
        values,
    ):

        values = np.clip(
            values,
            -50,
            50,
        )

        return (
            1
            /
            (
                1
                +
                np.exp(
                    -values
                )
            )
        )

    def find_best_stump(
        self,
        X,
        residuals,
    ):

        _, feature_count = (
            X.shape
        )

        best_error = np.inf

        best_feature = None
        best_threshold = None
        best_left = 0.0
        best_right = 0.0

        for feature_index in range(
            feature_count
        ):

            values = (
                X[
                    :,
                    feature_index
                ]
            )

            unique_values = np.unique(
                values
            )

            if len(
                unique_values
            ) < 2:

                continue

            thresholds = np.unique(
                np.quantile(
                    values,
                    np.linspace(
                        0.10,
                        0.90,
                        self.max_thresholds,
                    ),
                )
            )

            for threshold in (
                thresholds
            ):

                left_mask = (
                    values
                    <=
                    threshold
                )

                right_mask = (
                    ~left_mask
                )

                left_count = int(
                    left_mask.sum()
                )

                right_count = int(
                    right_mask.sum()
                )

                if (
                    left_count
                    <
                    self.min_leaf
                ):

                    continue

                if (
                    right_count
                    <
                    self.min_leaf
                ):

                    continue

                left_residual = (
                    residuals[
                        left_mask
                    ]
                )

                right_residual = (
                    residuals[
                        right_mask
                    ]
                )

                left_value = float(
                    np.mean(
                        left_residual
                    )
                )

                right_value = float(
                    np.mean(
                        right_residual
                    )
                )

                error = (

                    np.sum(
                        (
                            left_residual
                            -
                            left_value
                        )
                        ** 2
                    )

                    +

                    np.sum(
                        (
                            right_residual
                            -
                            right_value
                        )
                        ** 2
                    )
                )

                if (
                    error
                    <
                    best_error
                ):

                    best_error = (
                        error
                    )

                    best_feature = (
                        feature_index
                    )

                    best_threshold = (
                        threshold
                    )

                    best_left = (
                        left_value
                    )

                    best_right = (
                        right_value
                    )

        return {

            "feature":
                best_feature,

            "threshold":
                best_threshold,

            "left":
                best_left,

            "right":
                best_right,
        }

    def fit(
        self,
        X,
        y,
    ):

        X = np.asarray(
            X,
            dtype=float,
        )

        y = np.asarray(
            y,
            dtype=float,
        )

        positive_rate = np.clip(
            np.mean(y),
            1e-5,
            1 - 1e-5,
        )

        self.base_score = (
            np.log(
                positive_rate
                /
                (
                    1
                    -
                    positive_rate
                )
            )
        )

        current_score = np.full(
            len(y),
            self.base_score,
            dtype=float,
        )

        self.stumps = []

        for _ in range(
            self.n_estimators
        ):

            probability = (
                self.sigmoid(
                    current_score
                )
            )

            residuals = (
                y
                -
                probability
            )

            stump = (
                self.find_best_stump(
                    X,
                    residuals,
                )
            )

            if (
                stump[
                    "feature"
                ]
                is None
            ):

                break

            feature = (
                stump[
                    "feature"
                ]
            )

            threshold = (
                stump[
                    "threshold"
                ]
            )

            prediction = np.where(
                X[
                    :,
                    feature
                ]
                <=
                threshold,
                stump[
                    "left"
                ],
                stump[
                    "right"
                ],
            )

            current_score += (
                self.learning_rate
                *
                prediction
            )

            self.stumps.append(
                stump
            )

        return self

    def decision_function(
        self,
        X,
    ):

        if self.base_score is None:

            raise RuntimeError(
                "Gradient boosting model has not been fitted."
            )

        X = np.asarray(
            X,
            dtype=float,
        )

        score = np.full(
            len(X),
            self.base_score,
            dtype=float,
        )

        for stump in (
            self.stumps
        ):

            feature = (
                stump[
                    "feature"
                ]
            )

            threshold = (
                stump[
                    "threshold"
                ]
            )

            prediction = np.where(
                X[
                    :,
                    feature
                ]
                <=
                threshold,
                stump[
                    "left"
                ],
                stump[
                    "right"
                ],
            )

            score += (
                self.learning_rate
                *
                prediction
            )

        return score

    def predict_proba(
        self,
        X,
    ):

        score = (
            self.decision_function(
                X
            )
        )

        probability = (
            self.sigmoid(
                score
            )
        )

        return np.column_stack(
            [
                1 - probability,
                probability,
            ]
        )


# ======================================================================
# 16. MODEL TRAINING
# ======================================================================

print("\n" + "=" * 90)
print(
    "STEP 3 — MODEL TRAINING"
)
print("=" * 90)

logistic_model = (
    LogisticRegressionNumpy()
)

boosting_model = (
    GradientBoostingStumps()
)

print(
    "Training Logistic Regression..."
)

logistic_model.fit(
    X_train,
    y_train,
)

print(
    "Training Gradient Boosting..."
)

boosting_model.fit(
    X_train,
    y_train,
)

print(
    "Model training completed."
)


# ======================================================================
# 17. MODEL METRICS
# ======================================================================

def roc_curve_data(
    actual,
    probability,
):

    actual = np.asarray(
        actual
    ).astype(int)

    probability = np.asarray(
        probability
    ).astype(float)

    order = np.argsort(
        -probability
    )

    actual_sorted = (
        actual[
            order
        ]
    )

    probability_sorted = (
        probability[
            order
        ]
    )

    positive_count = max(
        1,
        int(
            np.sum(
                actual == 1
            )
        )
    )

    negative_count = max(
        1,
        int(
            np.sum(
                actual == 0
            )
        )
    )

    fpr = [0.0]
    tpr = [0.0]

    true_positive = 0
    false_positive = 0

    previous_probability = None

    for label, probability_value in zip(
        actual_sorted,
        probability_sorted,
    ):

        if (
            previous_probability
            is not None
            and
            probability_value
            !=
            previous_probability
        ):

            fpr.append(
                false_positive
                /
                negative_count
            )

            tpr.append(
                true_positive
                /
                positive_count
            )

        if label == 1:

            true_positive += 1

        else:

            false_positive += 1

        previous_probability = (
            probability_value
        )

    fpr.append(
        false_positive
        /
        negative_count
    )

    tpr.append(
        true_positive
        /
        positive_count
    )

    return (
        np.asarray(fpr),
        np.asarray(tpr),
    )


def area_under_curve(
    x,
    y,
):

    order = np.argsort(
        x
    )

    x_sorted = (
        x[
            order
        ]
    )

    y_sorted = (
        y[
            order
        ]
    )

    return float(
        np.trapz(
            y_sorted,
            x_sorted,
        )
    )


def precision_recall_data(
    actual,
    probability,
):

    actual = np.asarray(
        actual
    ).astype(int)

    probability = np.asarray(
        probability
    ).astype(float)

    order = np.argsort(
        -probability
    )

    sorted_actual = (
        actual[
            order
        ]
    )

    true_positive = np.cumsum(
        sorted_actual == 1
    )

    false_positive = np.cumsum(
        sorted_actual == 0
    )

    positive_count = max(
        1,
        int(
            np.sum(
                actual == 1
            )
        )
    )

    precision = (
        true_positive
        /
        np.maximum(
            1,
            true_positive
            +
            false_positive,
        )
    )

    recall = (
        true_positive
        /
        positive_count
    )

    precision = np.concatenate(
        [
            [1.0],
            precision,
        ]
    )

    recall = np.concatenate(
        [
            [0.0],
            recall,
        ]
    )

    return (
        precision,
        recall,
    )


def evaluate_model(
    actual,
    probability,
    threshold=0.05,
):

    actual = np.asarray(
        actual
    ).astype(int)

    probability = np.asarray(
        probability
    ).astype(float)

    prediction = (
        probability
        >=
        threshold
    ).astype(int)

    true_positive = int(
        np.sum(
            (
                actual == 1
            )
            &
            (
                prediction == 1
            )
        )
    )

    true_negative = int(
        np.sum(
            (
                actual == 0
            )
            &
            (
                prediction == 0
            )
        )
    )

    false_positive = int(
        np.sum(
            (
                actual == 0
            )
            &
            (
                prediction == 1
            )
        )
    )

    false_negative = int(
        np.sum(
            (
                actual == 1
            )
            &
            (
                prediction == 0
            )
        )
    )

    precision = (
        true_positive
        /
        max(
            1,
            true_positive
            +
            false_positive,
        )
    )

    recall = (
        true_positive
        /
        max(
            1,
            true_positive
            +
            false_negative,
        )
    )

    f1 = (
        2
        *
        precision
        *
        recall
        /
        max(
            1e-12,
            precision
            +
            recall,
        )
    )

    fpr, tpr = (
        roc_curve_data(
            actual,
            probability,
        )
    )

    precision_curve, recall_curve = (
        precision_recall_data(
            actual,
            probability,
        )
    )

    roc_auc_result = (
        area_under_curve(
            fpr,
            tpr,
        )
    )

    pr_auc_result = (
        area_under_curve(
            recall_curve,
            precision_curve,
        )
    )

    brier = float(
        np.mean(
            (
                probability
                -
                actual
            )
            ** 2
        )
    )

    probability_clipped = np.clip(
        probability,
        1e-12,
        1 - 1e-12,
    )

    log_loss_result = float(
        -
        np.mean(
            actual
            *
            np.log(
                probability_clipped
            )
            +
            (
                1
                -
                actual
            )
            *
            np.log(
                1
                -
                probability_clipped
            )
        )
    )

    ks = float(
        np.max(
            tpr
            -
            fpr
        )
    )

    return {

        "ROC_AUC":
            roc_auc_result,

        "PR_AUC":
            pr_auc_result,

        "Brier_Score":
            brier,

        "Log_Loss":
            log_loss_result,

        "KS":
            ks,

        "Precision":
            precision,

        "Recall":
            recall,

        "F1":
            f1,

        "True_Negatives":
            true_negative,

        "False_Positives":
            false_positive,

        "False_Negatives":
            false_negative,

        "True_Positives":
            true_positive,
    }


# ======================================================================
# 18. MODEL COMPARISON
# ======================================================================

logistic_validation_pd = (
    logistic_model
    .predict_proba(
        X_validation
    )[:, 1]
)

boosting_validation_pd = (
    boosting_model
    .predict_proba(
        X_validation
    )[:, 1]
)

logistic_metrics = (
    evaluate_model(
        y_validation,
        logistic_validation_pd,
    )
)

boosting_metrics = (
    evaluate_model(
        y_validation,
        boosting_validation_pd,
    )
)

validation_results_df = pd.DataFrame({

    "Logistic Regression":
        logistic_metrics,

    "Gradient Boosting":
        boosting_metrics,

}).T

print(
    "\nValidation model comparison:"
)

display(
    validation_results_df.round(
        5
    )
)

if (
    boosting_metrics[
        "PR_AUC"
    ]
    >
    logistic_metrics[
        "PR_AUC"
    ]
):

    selected_model_name = (
        "Gradient Boosting"
    )

else:

    selected_model_name = (
        "Logistic Regression"
    )

print(
    "\nSelected model:",
    selected_model_name,
)


# ======================================================================
# 19. FINAL MODEL REFIT
# ======================================================================

print(
    "\nRefitting selected model on training + validation data..."
)

X_trainval_raw = pd.concat(
    [
        X_train_raw,
        X_validation_raw,
    ],
    axis=0,
)

y_trainval = np.concatenate(
    [
        y_train,
        y_validation,
    ]
)

final_preprocessor = (
    fit_preprocessor(
        X_trainval_raw
    )
)

X_trainval = (
    transform_preprocessor(
        X_trainval_raw,
        final_preprocessor,
    )
)

X_test_final = (
    transform_preprocessor(
        X_test_raw,
        final_preprocessor,
    )
)

if (
    selected_model_name
    ==
    "Gradient Boosting"
):

    final_model = (
        GradientBoostingStumps()
    )

else:

    final_model = (
        LogisticRegressionNumpy()
    )

final_model.fit(
    X_trainval,
    y_trainval,
)

test_probability = (
    final_model
    .predict_proba(
        X_test_final
    )[:, 1]
)

test_metrics = (
    evaluate_model(
        y_test,
        test_probability,
    )
)

test_metrics_df = (
    pd.DataFrame(
        [
            test_metrics
        ]
    ).T
)

test_metrics_df.columns = [
    "Value"
]

print(
    "\nFINAL TEST PERFORMANCE"
)

display(
    test_metrics_df.round(
        5
    )
)


# ======================================================================
# 20. USEFUL GRAPH #1 — MODEL ROC
# ======================================================================

print(
    "\nGenerating model-performance visualization..."
)

test_fpr, test_tpr = (
    roc_curve_data(
        y_test,
        test_probability,
    )
)

plt.figure(
    figsize=(8, 6)
)

plt.plot(
    test_fpr,
    test_tpr,
    linewidth=2.5,
    label=(
        f"{selected_model_name} "
        f"(AUC = "
        f"{test_metrics['ROC_AUC']:.3f})"
    ),
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random benchmark",
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "Credit Risk Model — ROC Curve"
)

plt.legend()

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(
    CHART_DIR
    /
    "01_model_roc_curve.png",
    dpi=200,
)

plt.show()


# ======================================================================
# 21. SEC CLIENT
# ======================================================================

print("\n" + "=" * 90)
print(
    "STEP 4 — PUBLIC COMPANY DATA"
)
print("=" * 90)


class SECClient:

    annual_forms = {
        "10-K",
        "10-K/A",
        "20-F",
        "20-F/A",
        "40-F",
        "40-F/A",
    }

    def __init__(
        self,
        user_agent,
    ):

        self.user_agent = (
            user_agent
        )

        self.ticker_map = None

        self.response_cache = {}

    def get_json(
        self,
        url,
        retries=3,
    ):

        if url in self.response_cache:

            return (
                self.response_cache[
                    url
                ]
            )

        last_error = None

        for attempt in range(
            1,
            retries + 1,
        ):

            try:

                # --------------------------------------------------
                # Request identity encoding first.
                #
                # This prevents the gzip/UTF-8 error encountered
                # previously. If the server still returns gzip,
                # we decompress it explicitly.
                # --------------------------------------------------

                request = (
                    urllib.request.Request(
                        url,
                        headers={
                            "User-Agent":
                                self.user_agent,

                            "Accept":
                                "application/json",

                            "Accept-Encoding":
                                "identity",
                        },
                    )
                )

                with urllib.request.urlopen(
                    request,
                    timeout=90,
                ) as response:

                    raw = (
                        response.read()
                    )

                    content_encoding = (
                        response.headers.get(
                            "Content-Encoding",
                            "",
                        )
                        .lower()
                    )

                # --------------------------------------------------
                # Explicit gzip handling.
                # --------------------------------------------------

                if (
                    content_encoding
                    == "gzip"
                    or
                    (
                        len(raw) >= 2
                        and
                        raw[:2]
                        ==
                        b"\x1f\x8b"
                    )
                ):

                    raw = gzip.decompress(
                        raw
                    )

                text = raw.decode(
                    "utf-8",
                    errors="strict",
                )

                result = json.loads(
                    text
                )

                self.response_cache[
                    url
                ] = result

                return result

            except Exception as error:

                last_error = (
                    error
                )

                if attempt < retries:

                    time.sleep(
                        1.5
                        *
                        attempt
                    )

        raise RuntimeError(
            f"SEC request failed: {url}. "
            f"Error: {last_error}"
        )

    def get_ticker_map(
        self,
    ):

        if self.ticker_map is not None:

            return self.ticker_map

        data = self.get_json(
            SEC_TICKERS_URL
        )

        mapping = {}

        for record in data.values():

            ticker = (
                str(
                    record.get(
                        "ticker",
                        "",
                    )
                )
                .upper()
                .strip()
            )

            cik = record.get(
                "cik_str"
            )

            if ticker and cik:

                mapping[
                    ticker
                ] = (
                    str(
                        int(
                            cik
                        )
                    )
                    .zfill(
                        10
                    )
                )

        if not mapping:

            raise RuntimeError(
                "SEC ticker mapping is empty."
            )

        self.ticker_map = (
            mapping
        )

        return mapping

    def ticker_to_cik(
        self,
        ticker,
    ):

        ticker = (
            str(
                ticker
            )
            .upper()
            .strip()
        )

        mapping = (
            self.get_ticker_map()
        )

        if ticker not in mapping:

            raise ValueError(
                f"Ticker '{ticker}' "
                "was not found in SEC ticker map."
            )

        return mapping[
            ticker
        ]

    def company_facts(
        self,
        ticker,
    ):

        cik = (
            self.ticker_to_cik(
                ticker
            )
        )

        url = (
            "https://data.sec.gov/api/xbrl/"
            f"companyfacts/CIK{cik}.json"
        )

        return self.get_json(
            url
        )

    def latest_annual_fact(
        self,
        facts,
        tags,
    ):

        us_gaap = (
            facts
            .get(
                "facts",
                {},
            )
            .get(
                "us-gaap",
                {},
            )
        )

        for tag in tags:

            fact = (
                us_gaap.get(
                    tag
                )
            )

            if not fact:
                continue

            units = (
                fact.get(
                    "units",
                    {},
                )
            )

            # --------------------------------------------------
            # Prefer USD.
            # --------------------------------------------------

            if "USD" not in units:

                continue

            records = (
                units[
                    "USD"
                ]
            )

            candidates = []

            for record in records:

                form = (
                    record.get(
                        "form"
                    )
                )

                if (
                    form
                    not in
                    self.annual_forms
                ):

                    continue

                value = safe_float(
                    record.get(
                        "val"
                    )
                )

                end_date = (
                    record.get(
                        "end"
                    )
                )

                filed_date = (
                    record.get(
                        "filed",
                        end_date,
                    )
                )

                if (
                    np.isfinite(
                        value
                    )
                    and
                    end_date
                ):

                    candidates.append(
                        (
                            str(
                                filed_date
                                or
                                ""
                            ),

                            str(
                                end_date
                            ),

                            value,
                        )
                    )

            if candidates:

                candidates.sort()

                return float(
                    candidates[-1][2]
                )

        return np.nan

    def financials(
        self,
        ticker,
    ):

        facts = (
            self.company_facts(
                ticker
            )
        )

        # ----------------------------------------------------------
        # Multiple fallback tags are intentionally provided.
        # Different issuers can use different US-GAAP concepts.
        # ----------------------------------------------------------

        tags = {

            "assets": [
                "Assets",
            ],

            "liabilities": [
                "Liabilities",
            ],

            "current_assets": [
                "AssetsCurrent",
            ],

            "current_liabilities": [
                "LiabilitiesCurrent",
            ],

            "cash": [
                "CashAndCashEquivalentsAtCarryingValue",
                "CashAndCashEquivalentsFairValueDisclosure",
                "CashCashEquivalentsRestrictedCashAndRestrictedCashEquivalents",
            ],

            "equity": [
                "StockholdersEquity",
                "StockholdersEquityIncludingPortionAttributableToNoncontrollingInterest",
            ],

            "net_income": [
                "ProfitLoss",
                "NetIncomeLoss",
            ],

            "ebit": [
                "OperatingIncomeLoss",
            ],

            "sales": [
                "RevenueFromContractWithCustomerExcludingAssessedTax",
                "Revenues",
                "SalesRevenueNet",
            ],

            "gross_profit": [
                "GrossProfit",
            ],

            "cogs": [
                "CostOfRevenue",
                "CostOfGoodsAndServicesSold",
            ],

            "retained_earnings": [
                "RetainedEarningsAccumulatedDeficit",
            ],

            "interest_expense": [
                "InterestExpenseNonOperating",
                "InterestExpenseDebt",
                "InterestExpense",
            ],

            "long_term_debt": [
                "LongTermDebtNoncurrent",
                "LongTermDebt",
            ],
        }

        output = {}

        for field, tag_list in (
            tags.items()
        ):

            output[
                field
            ] = (
                self.latest_annual_fact(
                    facts,
                    tag_list,
                )
            )

        # ----------------------------------------------------------
        # Derived financial quantities.
        # ----------------------------------------------------------

        if (
            not np.isfinite(
                output[
                    "gross_profit"
                ]
            )
            and
            np.isfinite(
                output[
                    "sales"
                ]
            )
            and
            np.isfinite(
                output[
                    "cogs"
                ]
            )
        ):

            output[
                "gross_profit"
            ] = (
                output[
                    "sales"
                ]
                -
                output[
                    "cogs"
                ]
            )

        if (
            np.isfinite(
                output[
                    "assets"
                ]
            )
            and
            np.isfinite(
                output[
                    "current_assets"
                ]
            )
        ):

            output[
                "fixed_assets"
            ] = (
                output[
                    "assets"
                ]
                -
                output[
                    "current_assets"
                ]
            )

        else:

            output[
                "fixed_assets"
            ] = np.nan

        if np.isfinite(
            output[
                "interest_expense"
            ]
        ):

            output[
                "interest_expense"
            ] = abs(
                output[
                    "interest_expense"
                ]
            )

        return output


sec_client = SECClient(
    SEC_USER_AGENT
)


# ======================================================================
# 22. SEC FINANCIALS → MODEL FEATURES
# ======================================================================

def financials_to_features(
    financials,
):

    assets = (
        financials.get(
            "assets",
            np.nan,
        )
    )

    liabilities = (
        financials.get(
            "liabilities",
            np.nan,
        )
    )

    current_assets = (
        financials.get(
            "current_assets",
            np.nan,
        )
    )

    current_liabilities = (
        financials.get(
            "current_liabilities",
            np.nan,
        )
    )

    cash = (
        financials.get(
            "cash",
            np.nan,
        )
    )

    equity = (
        financials.get(
            "equity",
            np.nan,
        )
    )

    net_income = (
        financials.get(
            "net_income",
            np.nan,
        )
    )

    ebit = (
        financials.get(
            "ebit",
            np.nan,
        )
    )

    sales = (
        financials.get(
            "sales",
            np.nan,
        )
    )

    gross_profit = (
        financials.get(
            "gross_profit",
            np.nan,
        )
    )

    retained_earnings = (
        financials.get(
            "retained_earnings",
            np.nan,
        )
    )

    interest_expense = (
        financials.get(
            "interest_expense",
            np.nan,
        )
    )

    long_term_debt = (
        financials.get(
            "long_term_debt",
            np.nan,
        )
    )

    fixed_assets = (
        financials.get(
            "fixed_assets",
            np.nan,
        )
    )

    if (
        np.isfinite(
            current_assets
        )
        and
        np.isfinite(
            current_liabilities
        )
    ):

        working_capital = (
            current_assets
            -
            current_liabilities
        )

    else:

        working_capital = np.nan

    if (
        np.isfinite(
            equity
        )
        and
        np.isfinite(
            long_term_debt
        )
    ):

        constant_capital = (
            equity
            +
            long_term_debt
        )

    else:

        constant_capital = np.nan

    gross_margin = (
        safe_divide(
            gross_profit,
            sales,
        )
    )

    features = {

        "Attr1":
            safe_divide(
                net_income,
                assets,
            ),

        "Attr2":
            safe_divide(
                liabilities,
                assets,
            ),

        "Attr3":
            safe_divide(
                working_capital,
                assets,
            ),

        "Attr4":
            safe_divide(
                current_assets,
                current_liabilities,
            ),

        "Attr6":
            safe_divide(
                retained_earnings,
                assets,
            ),

        "Attr7":
            safe_divide(
                ebit,
                assets,
            ),

        "Attr8":
            safe_divide(
                equity,
                liabilities,
            ),

        "Attr9":
            safe_divide(
                sales,
                assets,
            ),

        "Attr10":
            safe_divide(
                equity,
                assets,
            ),

        "Attr19":
            gross_margin,

        "Attr23":
            safe_divide(
                net_income,
                sales,
            ),

        "Attr27":
            safe_divide(
                ebit,
                interest_expense,
            ),

        "Attr28":
            safe_divide(
                working_capital,
                fixed_assets,
            ),

        "Attr29":
            (
                np.log(
                    assets
                )
                if (
                    np.isfinite(
                        assets
                    )
                    and
                    assets > 0
                )
                else np.nan
            ),

        "Attr30":
            safe_divide(
                (
                    liabilities
                    -
                    cash
                )
                if (
                    np.isfinite(
                        liabilities
                    )
                    and
                    np.isfinite(
                        cash
                    )
                )
                else np.nan,
                sales,
            ),

        "Attr38":
            safe_divide(
                constant_capital,
                assets,
            ),

        "Attr42":
            safe_divide(
                ebit,
                sales,
            ),

        "Attr50":
            safe_divide(
                current_assets,
                liabilities,
            ),

        "Attr51":
            safe_divide(
                current_liabilities,
                assets,
            ),

        "Attr56":
            gross_margin,

        "Attr59":
            safe_divide(
                long_term_debt,
                equity,
            ),
    }

    return pd.DataFrame(
        [
            features
        ]
    )[
        MODEL_FEATURES
    ]


# ======================================================================
# 23. SYNTHETIC FALLBACK DATA
# ======================================================================
#
# Only used if SEC access temporarily fails.
# Every fallback company has its own illustrative capital structure.
# This ensures the project completes end-to-end without producing N/A
# values merely because of a temporary external API problem.
# ======================================================================

SYNTHETIC_FINANCIALS = {

    "MSFT": {

        "assets":
            619000.0,

        "liabilities":
            243000.0,

        "current_assets":
            191000.0,

        "current_liabilities":
            141000.0,

        "cash":
            94000.0,

        "equity":
            376000.0,

        "net_income":
            100000.0,

        "ebit":
            129000.0,

        "sales":
            281000.0,

        "gross_profit":
            193000.0,

        "cogs":
            88000.0,

        "retained_earnings":
            250000.0,

        "interest_expense":
            4500.0,

        "long_term_debt":
            76000.0,

        "fixed_assets":
            428000.0,
    },

    "AAPL": {

        "assets":
            365000.0,

        "liabilities":
            309000.0,

        "current_assets":
            151000.0,

        "current_liabilities":
            177000.0,

        "cash":
            62000.0,

        "equity":
            56000.0,

        "net_income":
            93000.0,

        "ebit":
            124000.0,

        "sales":
            394000.0,

        "gross_profit":
            184000.0,

        "cogs":
            210000.0,

        "retained_earnings":
            60000.0,

        "interest_expense":
            3900.0,

        "long_term_debt":
            93000.0,

        "fixed_assets":
            214000.0,
    },

    "AMZN": {

        "assets":
            650000.0,

        "liabilities":
            360000.0,

        "current_assets":
            190000.0,

        "current_liabilities":
            180000.0,

        "cash":
            86000.0,

        "equity":
            290000.0,

        "net_income":
            59000.0,

        "ebit":
            83000.0,

        "sales":
            716000.0,

        "gross_profit":
            340000.0,

        "cogs":
            376000.0,

        "retained_earnings":
            260000.0,

        "interest_expense":
            4200.0,

        "long_term_debt":
            138000.0,

        "fixed_assets":
            460000.0,
    },

    "WMT": {

        "assets":
            260000.0,

        "liabilities":
            162000.0,

        "current_assets":
            79000.0,

        "current_liabilities":
            92000.0,

        "cash":
            9000.0,

        "equity":
            98000.0,

        "net_income":
            20000.0,

        "ebit":
            28000.0,

        "sales":
            680000.0,

        "gross_profit":
            165000.0,

        "cogs":
            515000.0,

        "retained_earnings":
            62000.0,

        "interest_expense":
            5000.0,

        "long_term_debt":
            58000.0,

        "fixed_assets":
            181000.0,
    },
}


# ======================================================================
# 24. SCORE ALL SAMPLE COMPANIES
# ======================================================================

borrower_results = []

company_feature_store = {}

company_financial_store = {}

for ticker, company_info in (
    COMPANY_UNIVERSE.items()
):

    print(
        f"\nProcessing {ticker} — "
        f"{company_info['name']}"
    )

    try:

        # --------------------------------------------------------------
        # First attempt: real SEC data.
        # --------------------------------------------------------------

        financials = (
            sec_client.financials(
                ticker
            )
        )

        raw_features = (
            financials_to_features(
                financials
            )
        )

        # --------------------------------------------------------------
        # Validate SEC extraction.
        # We require several core fields to be present.
        # --------------------------------------------------------------

        core_fields = [

            "assets",
            "liabilities",
            "current_assets",
            "current_liabilities",
            "equity",
            "sales",
            "ebit",
        ]

        raw_values = [
            safe_float(
                financials.get(
                    field,
                    np.nan,
                )
            )

            for field
            in core_fields
        ]

        core_coverage = (
            sum(
                np.isfinite(
                    value
                )
                for value
                in raw_values
            )
            /
            len(core_fields)
        )

        if core_coverage < 0.75:

            raise ValueError(
                "SEC XBRL data coverage was insufficient "
                f"({core_coverage:.0%} of core fields)."
            )

        model_features = (
            transform_preprocessor(
                raw_features,
                final_preprocessor,
            )
        )

        pd_value = float(
            final_model
            .predict_proba(
                model_features
            )[0, 1]
        )

        source = (
            "SEC Company Facts API"
        )

    except Exception as sec_error:

        print(
            f"SEC extraction issue for {ticker}: "
            f"{sec_error}"
        )

        # --------------------------------------------------------------
        # Fallback.
        # --------------------------------------------------------------

        financials = (
            SYNTHETIC_FINANCIALS[
                ticker
            ].copy()
        )

        raw_features = (
            financials_to_features(
                financials
            )
        )

        model_features = (
            transform_preprocessor(
                raw_features,
                final_preprocessor,
            )
        )

        pd_value = float(
            final_model
            .predict_proba(
                model_features
            )[0, 1]
        )

        source = (
            "Illustrative synthetic fallback"
        )

    company_feature_store[
        ticker
    ] = raw_features.copy()

    company_financial_store[
        ticker
    ] = financials.copy()

    borrower_results.append({

        "Ticker":
            ticker,

        "Company":
            company_info[
                "name"
            ],

        "Sector":
            company_info[
                "sector"
            ],

        "Model_Implied_PD":
            pd_value,

        "Data_Source":
            source,
    })


borrower_results_df = pd.DataFrame(
    borrower_results
)

borrower_display_df = (
    borrower_results_df.copy()
)

borrower_display_df[
    "Model-Implied PD"
] = (
    borrower_display_df[
        "Model_Implied_PD"
    ].map(
        format_pct
    )
)

borrower_display_df = (
    borrower_display_df[
        [
            "Ticker",
            "Company",
            "Sector",
            "Model-Implied PD",
            "Data_Source",
        ]
    ]
)

print(
    "\nPublic-company borrower scores:"
)

display(
    borrower_display_df
)


# ======================================================================
# 25. PRIMARY BORROWER
# ======================================================================

primary_match = (
    borrower_results_df[
        borrower_results_df[
            "Ticker"
        ]
        ==
        PRIMARY_BORROWER
    ]
)

if primary_match.empty:

    raise RuntimeError(
        "Primary borrower was not scored."
    )

primary_pd = safe_float(
    primary_match[
        "Model_Implied_PD"
    ].iloc[0]
)

primary_financials = (
    company_financial_store[
        PRIMARY_BORROWER
    ]
)

primary_features = (
    company_feature_store[
        PRIMARY_BORROWER
    ]
)


# ======================================================================
# 26. USEFUL GRAPH #2 — BORROWER PD COMPARISON
# ======================================================================

print(
    "\nGenerating borrower comparison visualization..."
)

available_borrowers = (
    borrower_results_df
    .dropna(
        subset=[
            "Model_Implied_PD"
        ]
    )
    .copy()
)

plt.figure(
    figsize=(9, 5)
)

bars = plt.bar(
    available_borrowers[
        "Ticker"
    ],
    available_borrowers[
        "Model_Implied_PD"
    ]
    *
    100,
)

plt.xlabel(
    "Company"
)

plt.ylabel(
    "Model-Implied PD (%)"
)

plt.title(
    "Model-Implied Probability of Default — Sample Companies"
)

plt.grid(
    axis="y",
    alpha=0.20,
)

for bar in bars:

    height = (
        bar.get_height()
    )

    plt.text(
        bar.get_x()
        +
        bar.get_width()
        /
        2,
        height,
        f"{height:.2f}%",
        ha="center",
        va="bottom",
    )

plt.tight_layout()

plt.savefig(
    CHART_DIR
    /
    "02_borrower_pd_comparison.png",
    dpi=200,
)

plt.show()


# ======================================================================
# 27. ILLUSTRATIVE UNDERWRITING POLICY
# ======================================================================

UNDERWRITING_POLICY = pd.DataFrame({

    "max_pd": [

        0.02,
        0.05,
        0.10,
        0.20,
        np.inf,
    ],

    "spread_bps": [

        400,
        550,
        700,
        900,
        1100,
    ],

    "max_net_leverage": [

        5.00,
        4.50,
        4.00,
        3.25,
        2.50,
    ],

    "min_interest_coverage": [

        2.50,
        2.75,
        3.00,
        3.25,
        3.50,
    ],

    "policy_band": [

        "PD < 2%",

        "2% <= PD < 5%",

        "5% <= PD < 10%",

        "10% <= PD < 20%",

        "PD >= 20%",
    ],
})


# ======================================================================
# 28. HYPOTHETICAL DIRECT-LENDING DEAL
# ======================================================================
#
# These are illustrative assumptions for the project and are not
# presented as market quotes.
# ======================================================================

DEAL = {

    "base_rate":
        0.0500,

    "ebitda":
        150.0,

    "existing_debt":
        350.0,

    "cash":
        50.0,

    "proposed_facility":
        250.0,

    "existing_rate":
        0.0750,

    "lgd":
        0.45,

    "upfront_fee":
        0.015,

    "fee_years":
        5,
}


# ======================================================================
# 29. SELECT POLICY BAND
# ======================================================================

selected_policy = None

for _, row in (
    UNDERWRITING_POLICY.iterrows()
):

    if (
        primary_pd
        <
        row[
            "max_pd"
        ]
    ):

        selected_policy = (
            row
        )

        break

if selected_policy is None:

    selected_policy = (
        UNDERWRITING_POLICY.iloc[
            -1
        ]
    )


# ======================================================================
# 30. UNDERWRITING CALCULATIONS
# ======================================================================

spread_bps = float(
    selected_policy[
        "spread_bps"
    ]
)

coupon = (
    DEAL[
        "base_rate"
    ]
    +
    spread_bps
    /
    10000
)

maximum_net_leverage = float(
    selected_policy[
        "max_net_leverage"
    ]
)

minimum_interest_coverage = float(
    selected_policy[
        "min_interest_coverage"
    ]
)

existing_net_debt = (
    DEAL[
        "existing_debt"
    ]
    -
    DEAL[
        "cash"
    ]
)

maximum_total_net_debt = (
    DEAL[
        "ebitda"
    ]
    *
    maximum_net_leverage
)

maximum_facility = max(
    0.0,
    maximum_total_net_debt
    -
    existing_net_debt,
)

initial_net_leverage = (
    DEAL[
        "existing_debt"
    ]
    +
    DEAL[
        "proposed_facility"
    ]
    -
    DEAL[
        "cash"
    ]
)
/
max(
    DEAL[
        "ebitda"
    ],
    1e-12,
)

annual_interest = (
    DEAL[
        "existing_debt"
    ]
    *
    DEAL[
        "existing_rate"
    ]
    +
    DEAL[
        "proposed_facility"
    ]
    *
    coupon
)

interest_coverage = (
    safe_divide(
        DEAL[
            "ebitda"
        ],
        annual_interest,
    )
)

expected_loss = (
    primary_pd
    *
    DEAL[
        "lgd"
    ]
    *
    DEAL[
        "proposed_facility"
    ]
)

expected_loss_rate = (
    primary_pd
    *
    DEAL[
        "lgd"
    ]
)

annualized_fee_income = (
    DEAL[
        "proposed_facility"
    ]
    *
    DEAL[
        "upfront_fee"
    ]
    /
    DEAL[
        "fee_years"
    ]
)

risk_adjusted_yield = (
    coupon
    +
    DEAL[
        "upfront_fee"
    ]
    /
    DEAL[
        "fee_years"
    ]
    -
    expected_loss_rate
)

leverage_breach = (
    initial_net_leverage
    >
    maximum_net_leverage
)

coverage_breach = (
    interest_coverage
    <
    minimum_interest_coverage
)


# ======================================================================
# 31. DEAL ANALYSIS TABLE
# ======================================================================

deal_output = {

    "Borrower":
        COMPANY_UNIVERSE[
            PRIMARY_BORROWER
        ][
            "name"
        ],

    "Ticker":
        PRIMARY_BORROWER,

    "Business":
        COMPANY_UNIVERSE[
            PRIMARY_BORROWER
        ][
            "description"
        ],

    "Model":
        selected_model_name,

    "Model-Implied PD":
        format_pct(
            primary_pd
        ),

    "Underwriting Band":
        selected_policy[
            "policy_band"
        ],

    "Illustrative Spread":
        f"{spread_bps:.0f} bps",

    "Illustrative Coupon":
        format_pct(
            coupon
        ),

    "EBITDA":
        format_money(
            DEAL[
                "ebitda"
            ]
        ),

    "Proposed Facility":
        format_money(
            DEAL[
                "proposed_facility"
            ]
        ),

    "Maximum Facility":
        format_money(
            maximum_facility
        ),

    "Initial Net Leverage":
        format_multiple(
            initial_net_leverage
        ),

    "Maximum Net Leverage":
        format_multiple(
            maximum_net_leverage
        ),

    "Interest Coverage":
        format_multiple(
            interest_coverage
        ),

    "Minimum Interest Coverage":
        format_multiple(
            minimum_interest_coverage
        ),

    "LGD":
        format_pct(
            DEAL[
                "lgd"
            ]
        ),

    "Expected Loss":
        format_money(
            expected_loss
        ),

    "Annualized Fee Income":
        format_money(
            annualized_fee_income
        ),

    "Risk-Adjusted Yield":
        format_pct(
            risk_adjusted_yield
        ),

    "Leverage Covenant Breach":
        str(
            leverage_breach
        ),

    "Coverage Covenant Breach":
        str(
            coverage_breach
        ),
}

deal_df = pd.DataFrame({

    "Metric":
        list(
            deal_output.keys()
        ),

    "Value":
        list(
            deal_output.values()
        ),
})

print("\n" + "=" * 90)
print(
    "STEP 5 — DIRECT-LENDING UNDERWRITING ANALYSIS"
)
print("=" * 90)

display(
    deal_df
)


# ======================================================================
# 32. STRESS TEST
# ======================================================================

def run_stress_test(
    deal,
    policy_row,
):

    base_coupon = (
        deal[
            "base_rate"
        ]
        +
        float(
            policy_row[
                "spread_bps"
            ]
        )
        /
        10000
    )

    ebitda_shocks = [

        0.00,
        -0.10,
        -0.20,
        -0.30,
    ]

    rate_shocks = [

        0.00,
        0.01,
        0.02,
        0.03,
    ]

    rows = []

    for ebitda_shock in (
        ebitda_shocks
    ):

        stressed_ebitda = (
            deal[
                "ebitda"
            ]
            *
            (
                1
                +
                ebitda_shock
            )
        )

        for rate_shock in (
            rate_shocks
        ):

            stressed_coupon = (
                base_coupon
                +
                rate_shock
            )

            stressed_interest = (
                deal[
                    "existing_debt"
                ]
                *
                deal[
                    "existing_rate"
                ]
                +
                deal[
                    "proposed_facility"
                ]
                *
                stressed_coupon
            )

            stressed_coverage = (
                safe_divide(
                    stressed_ebitda,
                    stressed_interest,
                )
            )

            stressed_leverage = (
                safe_divide(
                    (
                        deal[
                            "existing_debt"
                        ]
                        +
                        deal[
                            "proposed_facility"
                        ]
                        -
                        deal[
                            "cash"
                        ]
                    ),
                    stressed_ebitda,
                )
            )

            rows.append({

                "EBITDA_Shock":
                    ebitda_shock,

                "Rate_Shock":
                    rate_shock,

                "Stressed_EBITDA":
                    stressed_ebitda,

                "Stressed_Coupon":
                    stressed_coupon,

                "Interest_Coverage":
                    stressed_coverage,

                "Net_Leverage":
                    stressed_leverage,
            })

    return pd.DataFrame(
        rows
    )


stress_df = (
    run_stress_test(
        DEAL,
        selected_policy,
    )
)

print(
    "\nStress-test results:"
)

display(
    stress_df.round(
        4
    )
)


# ======================================================================
# 33. USEFUL GRAPH #3 — STRESS HEATMAP
# ======================================================================
#
# This is included because stress testing is directly relevant to
# private-credit underwriting.
# ======================================================================

stress_heatmap = (
    stress_df
    .pivot(
        index="EBITDA_Shock",
        columns="Rate_Shock",
        values="Interest_Coverage",
    )
)

plt.figure(
    figsize=(9, 6)
)

heatmap_image = plt.imshow(
    stress_heatmap.values,
    aspect="auto",
    interpolation="nearest",
)

plt.colorbar(
    heatmap_image,
    label="Interest Coverage"
)

plt.xticks(
    range(
        len(
            stress_heatmap.columns
        )
    ),
    [
        f"+{x:.0%}"
        for x
        in
        stress_heatmap.columns
    ],
)

plt.yticks(
    range(
        len(
            stress_heatmap.index
        )
    ),
    [
        f"{x:.0%}"
        for x
        in
        stress_heatmap.index
    ],
)

plt.xlabel(
    "Interest Rate Shock"
)

plt.ylabel(
    "EBITDA Shock"
)

plt.title(
    "Interest Coverage Stress Test"
)

for row_number in range(
    len(
        stress_heatmap.index
    )
):

    for column_number in range(
        len(
            stress_heatmap.columns
        )
    ):

        value = (
            stress_heatmap.iloc[
                row_number,
                column_number
            ]
        )

        plt.text(
            column_number,
            row_number,
            f"{value:.2f}x",
            ha="center",
            va="center",
            fontsize=10,
        )

plt.tight_layout()

plt.savefig(
    CHART_DIR
    /
    "03_interest_coverage_stress.png",
    dpi=200,
)

plt.show()


# ======================================================================
# 34. BORROWER FEATURE EXPORT
# ======================================================================

borrower_features_export = (
    primary_features.T
    .reset_index()
)

borrower_features_export.columns = [
    "Feature",
    "Value",
]

borrower_features_export[
    "Description"
] = (
    borrower_features_export[
        "Feature"
    ]
    .map(
        FEATURE_LABELS
    )
)


# ======================================================================
# 35. SAVE CSV OUTPUTS
# ======================================================================

metrics_export = (
    test_metrics_df
    .reset_index()
)

metrics_export.columns = [
    "Metric",
    "Value",
]

metrics_export.to_csv(
    OUTPUT_DIR
    /
    "model_metrics.csv",
    index=False,
)

borrower_results_df.to_csv(
    OUTPUT_DIR
    /
    "borrower_scores.csv",
    index=False,
)

borrower_features_export.to_csv(
    OUTPUT_DIR
    /
    "borrower_features.csv",
    index=False,
)

deal_df.to_csv(
    OUTPUT_DIR
    /
    "deal_analysis.csv",
    index=False,
)

stress_df.to_csv(
    OUTPUT_DIR
    /
    "stress_test.csv",
    index=False,
)


# ======================================================================
# 36. SQLITE DATABASE
# ======================================================================

print("\n" + "=" * 90)
print(
    "STEP 6 — SQL DATABASE"
)
print("=" * 90)

database_path = (
    SQL_DIR
    /
    "credit_risk.db"
)

with sqlite3.connect(
    database_path
) as connection:

    # --------------------------------------------------------------
    # IMPORTANT:
    # target is called default_flag instead of default.
    # --------------------------------------------------------------

    historical_df.to_sql(
        "historical_bankruptcy_data",
        connection,
        if_exists="replace",
        index=False,
    )

    borrower_results_df.to_sql(
        "public_company_scores",
        connection,
        if_exists="replace",
        index=False,
    )

    borrower_features_export.to_sql(
        "primary_borrower_features",
        connection,
        if_exists="replace",
        index=False,
    )

    deal_df.to_sql(
        "deal_analysis",
        connection,
        if_exists="replace",
        index=False,
    )

    stress_df.to_sql(
        "stress_test",
        connection,
        if_exists="replace",
        index=False,
    )

    metrics_export.to_sql(
        "model_metrics",
        connection,
        if_exists="replace",
        index=False,
    )

    # --------------------------------------------------------------
    # SQL validation.
    # --------------------------------------------------------------

    sql_validation_query = """
        SELECT
            COUNT(*) AS observations,
            SUM(default_flag) AS defaults,
            AVG(default_flag) AS default_rate
        FROM historical_bankruptcy_data;
    """

    sql_validation = (
        pd.read_sql_query(
            sql_validation_query,
            connection,
        )
    )

print(
    "SQL validation:"
)

display(
    sql_validation
)


# ======================================================================
# 37. MODEL ARTIFACT
# ======================================================================

model_artifact = {

    "project":
        PROJECT_TITLE,

    "selected_model":
        selected_model_name,

    "features":
        MODEL_FEATURES,

    "feature_labels":
        FEATURE_LABELS,

    "primary_borrower":
        PRIMARY_BORROWER,

    "primary_borrower_name":
        COMPANY_UNIVERSE[
            PRIMARY_BORROWER
        ][
            "name"
        ],

    "model_implied_pd":
        float(
            primary_pd
        ),

    "test_metrics":
        {
            key:
                float(value)

            for key, value
            in test_metrics.items()
        },

    "illustrative_spread_bps":
        spread_bps,

    "illustrative_coupon":
        coupon,

    "maximum_facility":
        maximum_facility,

    "expected_loss":
        expected_loss,

    "risk_adjusted_yield":
        risk_adjusted_yield,
}

if (
    selected_model_name
    ==
    "Logistic Regression"
):

    model_artifact[
        "intercept"
    ] = float(
        final_model.weights[0]
    )

    model_artifact[
        "coefficients"
    ] = [
        float(
            value
        )
        for value
        in final_model.weights[
            1:
        ]
    ]

else:

    model_artifact[
        "base_score"
    ] = float(
        final_model.base_score
    )

    model_artifact[
        "learning_rate"
    ] = float(
        final_model.learning_rate
    )

    model_artifact[
        "stumps"
    ] = [

        {

            "feature":
                int(
                    stump[
                        "feature"
                    ]
                ),

            "threshold":
                float(
                    stump[
                        "threshold"
                    ]
                ),

            "left":
                float(
                    stump[
                        "left"
                    ]
                ),

            "right":
                float(
                    stump[
                        "right"
                    ]
                ),
        }

        for stump
        in final_model.stumps
    ]

model_path = (
    MODEL_DIR
    /
    "final_pd_model.json"
)

model_path.write_text(
    json.dumps(
        model_artifact,
        indent=2,
    ),
    encoding="utf-8",
)


# ======================================================================
# 38. EXCEL CREDIT MEMO
# ======================================================================

print(
    "\nGenerating Excel credit memo..."
)

excel_path = (
    OUTPUT_DIR
    /
    "credit_memo.xlsx"
)

workbook = Workbook()

default_sheet = (
    workbook.active
)

workbook.remove(
    default_sheet
)

# --------------------------------------------------------------
# Excel formatting.
# --------------------------------------------------------------

title_fill = PatternFill(
    fill_type="solid",
    fgColor="17365D",
)

header_fill = PatternFill(
    fill_type="solid",
    fgColor="D9E1F2",
)

title_font = Font(
    color="FFFFFF",
    bold=True,
    size=15,
)

header_font = Font(
    bold=True,
)


def add_dataframe_sheet(
    workbook,
    sheet_name,
    dataframe,
    title,
):

    worksheet = (
        workbook.create_sheet(
            title=sheet_name
        )
    )

    worksheet["A1"] = (
        title
    )

    worksheet["A1"].fill = (
        title_fill
    )

    worksheet["A1"].font = (
        title_font
    )

    worksheet.merge_cells(
        start_row=1,
        start_column=1,
        end_row=1,
        end_column=max(
            2,
            len(
                dataframe.columns
            ),
        ),
    )

    header_row = 3

    for column_number, column_name in enumerate(
        dataframe.columns,
        start=1,
    ):

        cell = worksheet.cell(
            header_row,
            column_number,
            column_name,
        )

        cell.fill = (
            header_fill
        )

        cell.font = (
            header_font
        )

    for row_number, row_values in enumerate(
        dataframe.itertuples(
            index=False
        ),
        start=4,
    ):

        for column_number, value in enumerate(
            row_values,
            start=1,
        ):

            worksheet.cell(
                row_number,
                column_number,
                value,
            )

    for column_number in range(
        1,
        len(
            dataframe.columns
        )
        +
        1,
    ):

        max_length = 0

        for row in (
            worksheet.iter_rows(
                min_col=column_number,
                max_col=column_number,
            )
        ):

            for cell in row:

                if cell.value is None:
                    continue

                max_length = max(
                    max_length,
                    len(
                        str(
                            cell.value
                        )
                    ),
                )

        worksheet.column_dimensions[
            get_column_letter(
                column_number
            )
        ].width = min(
            60,
            max(
                12,
                max_length + 2,
            ),
        )

    worksheet.freeze_panes = (
        "A4"
    )


# --------------------------------------------------------------
# Executive summary sheet.
# --------------------------------------------------------------

executive_summary_df = pd.DataFrame({

    "Metric": [

        "Company",

        "Ticker",

        "Sector",

        "Business Description",

        "Data Source",

        "Selected Model",

        "Model-Implied PD",

        "Underwriting Band",

        "Illustrative Spread",

        "Illustrative Coupon",

        "Proposed Facility",

        "Maximum Facility",

        "Initial Net Leverage",

        "Maximum Net Leverage",

        "Interest Coverage",

        "Minimum Interest Coverage",

        "LGD",

        "Expected Loss",

        "Risk-Adjusted Yield",

        "Leverage Covenant Breach",

        "Coverage Covenant Breach",
    ],

    "Value": [

        COMPANY_UNIVERSE[
            PRIMARY_BORROWER
        ][
            "name"
        ],

        PRIMARY_BORROWER,

        COMPANY_UNIVERSE[
            PRIMARY_BORROWER
        ][
            "sector"
        ],

        COMPANY_UNIVERSE[
            PRIMARY_BORROWER
        ][
            "description"
        ],

        primary_match[
            "Data_Source"
        ].iloc[0],

        selected_model_name,

        format_pct(
            primary_pd
        ),

        selected_policy[
            "policy_band"
        ],

        f"{spread_bps:.0f} bps",

        format_pct(
            coupon
        ),

        format_money(
            DEAL[
                "proposed_facility"
            ]
        ),

        format_money(
            maximum_facility
        ),

        format_multiple(
            initial_net_leverage
        ),

        format_multiple(
            maximum_net_leverage
        ),

        format_multiple(
            interest_coverage
        ),

        format_multiple(
            minimum_interest_coverage
        ),

        format_pct(
            DEAL[
                "lgd"
            ]
        ),

        format_money(
            expected_loss
        ),

        format_pct(
            risk_adjusted_yield
        ),

        str(
            leverage_breach
        ),

        str(
            coverage_breach
        ),
    ],
})


add_dataframe_sheet(
    workbook,
    "Executive Summary",
    executive_summary_df,
    "PRIVATE CREDIT — EXECUTIVE SUMMARY",
)

add_dataframe_sheet(
    workbook,
    "Company Universe",
    company_overview_df,
    "SAMPLE PUBLIC COMPANY UNIVERSE",
)

add_dataframe_sheet(
    workbook,
    "Borrower Scores",
    borrower_results_df,
    "PUBLIC COMPANY PD SCORES",
)

add_dataframe_sheet(
    workbook,
    "Model Metrics",
    metrics_export,
    "CREDIT MODEL PERFORMANCE",
)

add_dataframe_sheet(
    workbook,
    "Borrower Features",
    borrower_features_export,
    "PRIMARY BORROWER FEATURES",
)

add_dataframe_sheet(
    workbook,
    "Deal Analysis",
    deal_df,
    "DIRECT-LENDING UNDERWRITING",
)

add_dataframe_sheet(
    workbook,
    "Stress Test",
    stress_df,
    "STRESS TESTING",
)

add_dataframe_sheet(
    workbook,
    "Underwriting Policy",
    UNDERWRITING_POLICY,
    "ILLUSTRATIVE UNDERWRITING POLICY",
)

# Keep Excel manageable by storing only the first 1,000 historical rows.
add_dataframe_sheet(
    workbook,
    "Historical Sample",
    historical_df.head(
        1000
    ),
    "HISTORICAL BANKRUPTCY SAMPLE",
)

workbook.save(
    excel_path
)

print(
    "Excel credit memo saved:"
)

print(
    excel_path
)


# ======================================================================
# 39. README
# ======================================================================

readme_text = f"""
# {PROJECT_TITLE}

## Overview

{PROJECT_DESCRIPTION.strip()}

## Sample Companies

### Microsoft Corporation (MSFT)
Technology company with businesses spanning software, cloud computing,
enterprise technology and digital services.

### Apple Inc. (AAPL)
Technology company that develops consumer devices, software and digital
services.

### Amazon.com, Inc. (AMZN)
Global company operating across e-commerce, cloud computing, AI,
digital services and logistics.

### Walmart Inc. (WMT)
Global omnichannel retailer operating stores, clubs and e-commerce
channels.

## Primary Borrower

{COMPANY_UNIVERSE[PRIMARY_BORROWER]["name"]}

## Selected Model

{selected_model_name}

## Model-Implied PD

{primary_pd:.4%}

## Illustrative Direct-Lending Analysis

Proposed facility:
${DEAL["proposed_facility"]:.1f}mm

Maximum facility:
${maximum_facility:.1f}mm

Initial net leverage:
{initial_net_leverage:.2f}x

Interest coverage:
{interest_coverage:.2f}x

Expected loss:
${expected_loss:.2f}mm

Risk-adjusted yield:
{risk_adjusted_yield:.2%}

## Important Limitation

The historical bankruptcy training population and current SEC public-
company scoring population are different populations and time periods.

The resulting PD is therefore a model-implied research estimate rather
than a production-calibrated U.S. private-credit default probability.

## Outputs

- outputs/credit_memo.xlsx
- outputs/model_metrics.csv
- outputs/borrower_scores.csv
- outputs/borrower_features.csv
- outputs/deal_analysis.csv
- outputs/stress_test.csv
- outputs/charts/
- models/final_pd_model.json
- sql/credit_risk.db

## Future Enhancements

- Survival analysis
- LGD model
- EAD model
- Sector-specific models
- SHAP explainability
- Macroeconomic variables
- Market-implied credit risk
- Walk-forward validation
- API deployment
"""

(
    PROJECT_ROOT
    /
    "README.md"
).write_text(
    readme_text.strip(),
    encoding="utf-8",
)


# ======================================================================
# 40. REQUIREMENTS.TXT
# ======================================================================

requirements_text = """
numpy
pandas
openpyxl
""".strip()

(
    PROJECT_ROOT
    /
    "requirements.txt"
).write_text(
    requirements_text,
    encoding="utf-8",
)


# ======================================================================
# 41. FINAL CLOSING ANALYSIS
# ======================================================================

print("\n" + "=" * 90)
print(
    "STEP 7 — FINAL CREDIT ANALYSIS"
)
print("=" * 90)

closing_text = f"""
FINAL CREDIT ANALYSIS

Company:
{COMPANY_UNIVERSE[PRIMARY_BORROWER]["name"]}

Ticker:
{PRIMARY_BORROWER}

Business:
{COMPANY_UNIVERSE[PRIMARY_BORROWER]["description"]}

Model:
{selected_model_name}

Model-Implied Probability of Default:
{primary_pd:.2%}

Underwriting Band:
{selected_policy["policy_band"]}

Illustrative Pricing:
{spread_bps:.0f} bps spread over the assumed
{DEAL["base_rate"]:.2%} base rate, resulting in an
illustrative coupon of {coupon:.2%}.

Hypothetical Facility:
${DEAL["proposed_facility"]:.1f}mm

Maximum Facility Under the Illustrative Leverage Policy:
${maximum_facility:.1f}mm

Initial Net Leverage:
{initial_net_leverage:.2f}x

Maximum Net Leverage:
{maximum_net_leverage:.2f}x

Interest Coverage:
{interest_coverage:.2f}x

Minimum Interest Coverage:
{minimum_interest_coverage:.2f}x

Illustrative Expected Loss:
${expected_loss:.2f}mm

Illustrative Risk-Adjusted Yield:
{risk_adjusted_yield:.2%}

Covenant Checks:
Leverage covenant breach = {leverage_breach}
Interest coverage covenant breach = {coverage_breach}

The analysis therefore takes the selected company's financial profile,
passes it through the historical credit-risk model, and translates the
result into a hypothetical direct-lending structure.

The stress-testing section additionally measures how interest coverage
and leverage change when EBITDA declines and interest rates increase.

The final outputs have been saved to the project directory as an Excel
credit memo, CSV datasets, a SQLite database, model artifacts and
selected underwriting charts.

The PD should be interpreted as a model-implied research estimate
because the historical bankruptcy training population is different
from the current SEC public-company scoring population.
"""

print(
    closing_text.strip()
)


# ======================================================================
# 42. FINAL OUTPUT AUDIT
# ======================================================================

print("\n" + "=" * 90)
print(
    "FINAL PROJECT AUDIT"
)
print("=" * 90)

expected_files = [

    PROJECT_ROOT
    /
    "README.md",

    PROJECT_ROOT
    /
    "requirements.txt",

    OUTPUT_DIR
    /
    "credit_memo.xlsx",

    OUTPUT_DIR
    /
    "model_metrics.csv",

    OUTPUT_DIR
    /
    "borrower_scores.csv",

    OUTPUT_DIR
    /
    "borrower_features.csv",

    OUTPUT_DIR
    /
    "deal_analysis.csv",

    OUTPUT_DIR
    /
    "stress_test.csv",

    MODEL_DIR
    /
    "final_pd_model.json",

    SQL_DIR
    /
    "credit_risk.db",

    CHART_DIR
    /
    "01_model_roc_curve.png",

    CHART_DIR
    /
    "02_borrower_pd_comparison.png",

    CHART_DIR
    /
    "03_interest_coverage_stress.png",
]

audit_rows = []

for file_path in (
    expected_files
):

    exists = (
        file_path.exists()
    )

    size_kb = (

        round(
            file_path.stat().st_size
            /
            1024,
            2,
        )

        if exists

        else np.nan
    )

    audit_rows.append({

        "File":
            str(
                file_path
            ),

        "Exists":
            exists,

        "Size_KB":
            size_kb,
    })

audit_df = pd.DataFrame(
    audit_rows
)

display(
    audit_df
)

missing_outputs = (
    audit_df.loc[
        ~audit_df[
            "Exists"
        ],
        "File",
    ]
    .tolist()
)

if missing_outputs:

    print(
        "\nWARNING — Missing output files:"
    )

    for item in (
        missing_outputs
    ):

        print(
            "-",
            item,
        )

else:

    print(
        "\nALL EXPECTED OUTPUTS WERE CREATED."
    )


# ======================================================================
# 43. FINAL CONSOLE SUMMARY
# ======================================================================

print("\n" + "=" * 90)
print(
    "CREDIT RISK PROJECT COMPLETE"
)
print("=" * 90)

print(
    f"Company:             "
    f"{COMPANY_UNIVERSE[PRIMARY_BORROWER]['name']}"
)

print(
    f"Ticker:              "
    f"{PRIMARY_BORROWER}"
)

print(
    f"Model:               "
    f"{selected_model_name}"
)

print(
    f"Model-Implied PD:    "
    f"{format_pct(primary_pd)}"
)

print(
    f"Spread:              "
    f"{spread_bps:.0f} bps"
)

print(
    f"Coupon:              "
    f"{format_pct(coupon)}"
)

print(
    f"Proposed Facility:   "
    f"{format_money(DEAL['proposed_facility'])}"
)

print(
    f"Maximum Facility:    "
    f"{format_money(maximum_facility)}"
)

print(
    f"Net Leverage:        "
    f"{format_multiple(initial_net_leverage)}"
)

print(
    f"Interest Coverage:   "
    f"{format_multiple(interest_coverage)}"
)

print(
    f"Expected Loss:       "
    f"{format_money(expected_loss)}"
)

print(
    f"Risk-Adjusted Yield: "
    f"{format_pct(risk_adjusted_yield)}"
)

print()
print(
    "Project directory:"
)

print(
    PROJECT_ROOT
)

print()
print(
    "Excel credit memo:"
)

print(
    excel_path
)

print()
print(
    "SQLite database:"
)

print(
    database_path
)

print()
print(
    "Model artifact:"
)

print(
    model_path
)

print()
print(
    "Charts:"
)

print(
    CHART_DIR
)

print()
print("=" * 90)
print(
    "END OF PROJECT"
)
print("=" * 90)
```
